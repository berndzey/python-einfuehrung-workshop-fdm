## Zugriff auf Spalten und Zeilen
- mit Spaltenname
- mit Teilmenge der Spaltennamen
- bedingter Zugriff
- Tabelle in neuer Datei speichern

```python
# Spalten-Namen ausgeben
data.columns
```

```python
# Zugriff auf Spalte mit Spaltenname, z.B. "Name"
data["Name"]
```

```python
# Spalte "Age"
data["Age"]
```

```python
# Teilmenge der Spalten "Name", "Age", "Cabin"
# -> Teilmenge als Liste mit []
data[ ["Name", "Age", "Cabin"] ]
```

```python
# Filtern
# Mathematische Bedingungen >, <, =, >=, <=
# Hier: Bedingung Age >= 66
# Erklärung: liefert True-False-Vektor für welche Zeilen die Bedingung wahr/falsch ist
data["Age"] >= 66
```

```python
# Bedingung Age >= 33
data["Age"] >= 33
```

```python
# Auswahl nach Kriterien (Vektor als Filter)
# zeigt nur Zeilen mit Age >= 66 an
data[ data["Age"] >= 66 ]
```

```python
# Komplexere Auswahl nach Kriterien
# hier: Age >= 66 UND (Embarked == S ODER Embarked == Q)
data[ (data["Age"] >= 66) & ((data["Embarked"] == "S") | (data["Embarked"] == "Q"))]
```

```python
# Neue Tabelle in data3 speichern (zuweisen)
data3 = data[ (data["Age"] >= 66) & ((data["Embarked"] == "S") | (data["Embarked"] == "Q"))]
```

```python
# data3 ausgeben
data3
```

```python
# Indizes von data3 ausgeben
data3.index
```

```python
# Index von data3 neu setzen nit reset_index()
# drop=True: alten Index löschen (sonst wird eine neue Spalte "index" hinzugefügt)
data3.reset_index(drop=True)
```

```python
# data3 ausgeben
data3
```

```python
# Beobachtung: data3 ist unverändert 
# Begründung: man benötigt eine Zuweisung, sonst ändert sich data3 nicht!
data3 = data3.reset_index(drop=True)
```

```python
# data3 ausgeben
data3
```

```python
# Neue Tabelle data3 als csv und Excel speichern
# mit Trennzeichen ";" und Dezimalzeichen ","
# Dafür: to_csv() und analog zu read_csv()
data3.to_csv("test.csv", sep=";", decimal=",")
data3.to_excel("test.xlsx")
```

## Weitere Themen

```python
# Sortierung nach einer Spalte, hier: Alter
# Zuweisung nicht vergessen, sonst bleibt Tabelle unverändert!
data = data.sort_values(by="Age")
# Hinweis: auch mehrere Spalten bei "by" möglich, z.B.: by=["Age","Pclass"]
```

```python
# data anzeigen
data
```

```python
# Index resetten
# Zuweisung nicht vergessen...
data = data.reset_index(drop=True)
```

```python
# data anzeigen
data
```

```python
# Manche Elemente sind ohne Wert (NaN; Not a Number)
# True/False-Vektor/Matrix für die NaN-Werte mit isna()
data[data["Age"].isna()]
```

```python
# Nicht-NaN-Werte mit notna()
# identisch: notnull()
data[data["Age"].notna()]
```

```python
# Zeilen mit nicht-leeren "Cabin"-Zellen 
data[data["Cabin"].notna()]
```

```python
# Zeilen mit ausschließlich nicht-leeren Zellen
data.dropna()
```

```python
# NaN-Werte in Cabin-Spalte ersetzen mit fillna()
# Hier: mit "noCabin" ersetzen
data["Cabin"] = data["Cabin"].fillna("noCabin")
```

```python
# data anzeigen
data
```

```python
# Wert in Spalte ersetzen
# Hier: "noCabin" mit "0"
data["Cabin"] = data["Cabin"].replace("noCabin", 0)
```

```python
# data anzeigen
data
```

```python
# Alle (eindeutigen) Werte einer Spalte ausgeben
data["Pclass"].unique()
```
